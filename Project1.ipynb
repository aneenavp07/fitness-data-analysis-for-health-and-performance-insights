```python
for col in d:
    plt.figure(figsize=(6,4))
    plt.hist(d[col].dropna(), bins=30, color='orange', edgecolor='black')
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
import warnings
warnings.filterwarnings('ignore')
plt.figure(figsize=(6,4))
sns.barplot(x='Gender', y='Calories_Burned', data=d, palette='coolwarm', estimator='mean')
plt.title("Average Calories Burned by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Calories Burned")
plt.show()
```

```python
categorical_cols = d.select_dtypes(include='object').columns

for col in categorical_cols:
    plt.figure(figsize=(6,4))
    sns.countplot(x=col, data=d, palette='viridis')
    plt.title(f"Count of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=30)
    plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Workout_Type', y='Calories_Burned', data=d, palette='Set3')
plt.title("Calories Burned by Workout Type")
plt.xlabel("Workout Type")
plt.ylabel("Calories Burned")
plt.xticks(rotation=30)
plt.show()
```

```python
# Boxplot: Calories Burned by Workout Type
plt.figure(figsize=(8,5))
sns.boxplot(x='Workout_Type', y='Calories_Burned', data=d, palette='Set3')
plt.title("Calories Burned by Workout Type")
plt.show()

# Barplot: Average Calories by Gender
plt.figure(figsize=(6,4))
sns.barplot(x='Gender', y='Calories_Burned', data=d, estimator='mean', palette='pastel')
plt.title("Average Calories Burned by Gender")
plt.show()

```

```python
#1.What is the age distribution of users in the dataset?
sns.histplot(d['Age'], bins=30, kde=True, color='skyblue')
plt.title("Age Distribution of Users")
plt.show()
```

```python
#2.What is the BMI range of most users?
sns.histplot(d['BMI'], bins=30, kde=True, color='orange')
plt.title("BMI Distribution")
plt.show()
```

```python
#3.How many calories do users typically burn in a session?
plt.hist(d['Calories_Burned'], bins=30, color='lightgreen', edgecolor='black')
plt.title("Calories Burned Distribution")
plt.show()

```

```python
#4.What is the gender distribution among users?
sns.countplot(x='Gender', data=d, palette='pastel')
plt.title("Gender Distribution")
plt.show()
```

```python
#5.Which workout types are most commonly performed?
sns.countplot(x='Workout_Type', data=d, palette='viridis')
plt.title("Most Common Workout Types")
plt.xticks(rotation=30)
plt.show()
```

```python
#6.What is the relationship between Max BPM and Calories Burned?
sns.scatterplot(x='Max_BPM', y='Calories_Burned', data=d, color='crimson')
plt.title("Max BPM vs Calories Burned")
plt.show()

```

```python
#7.How is protein intake distributed among users?
sns.histplot(d['protein_per_kg'], kde=True, color='purple')
plt.title("Protein Intake Distribution")
plt.show()
```

```python
#8 Is there any relationship between protein intake and calories burned?
sns.scatterplot(x='protein_per_kg', y='Calories_Burned', data=d, color='darkgreen')
plt.title("Protein Intake vs Calories Burned")
plt.show()
```

```python
#9.How does Resting BPM differ across workout types?
sns.boxplot(x='Workout_Type', y='Resting_BPM', data=d, palette='magma')
plt.title("Resting BPM by Workout Type")
plt.show()
```
